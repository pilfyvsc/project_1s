```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.plot(x,y)
plt.xlabel("X Axis")
plt.ylabel('Y Axis')
plt.title("Simple Line Plot")
plt.show() # NOT required in Jupyter Notebooks

```

Module	        Purpose

pyplot - Creating graphs and charts easily

animation -  animated graphs

patches - Creating shapes such as circles, rectangles and polygons

colors - Working with colours and colour maps

dates - Handling dates and times on graphs

ticker - Controlling axis tick marks and labels

gridspec - Controlling the arrangement of multiple graphs

image - Reading and displaying images

figure - Creating and managing figures

axes - Controlling individual graph areas

```python
# figure and axes are individual modules
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_title('Line Plot (Object-Oriented)')
plt.show()
```
